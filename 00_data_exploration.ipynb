```python
import pandas as pd

# 修改为实际路径
data_path = '/root/flight_data.pkl'

print("正在加载云端数据...")
data = pd.read_pickle(data_path)
print("✅ 数据加载成功！")

print("数据类型:", type(data))
if isinstance(data, dict):
    print("数据是一个字典，包含以下键（keys）:")
    print(data.keys())
else:
    print("数据前5行预览:")
    print(data.head())
```

```python
# 查看字典里第一个键对应的值
first_key = list(data.keys())[0]
print(f"第一个键是: {first_key}")
print(f"对应值的类型: {type(data[first_key])}")

# 如果是DataFrame，查看其形状和前几行
if isinstance(data[first_key], pd.DataFrame):
    print(f"DataFrame形状: {data[first_key].shape}")
    print("前5行预览:")
    print(data[first_key].head())
else:
    # 如果是其他类型，直接打印
    print(data[first_key])
```

```python
# 加载 flight_header.csv
header_path = '/root/flight_header.csv'  # 假设它也在 /root/ 下，如果不在请调整路径
header_df = pd.read_csv(header_path)
print("header_df 的形状:", header_df.shape)
print("\n前10行预览:")
print(header_df.head(10))
print("\n所有列名:")
print(header_df.columns.tolist())
```

```python
# 1. 确保 header_df 中 Master Index 是 int 类型
header_df['Master Index'] = header_df['Master Index'].astype(int)

# 2. 将数据字典转换为列表，便于后续处理
# 注意：data 字典的键是航班ID，值是 NumPy 数组
flight_ids = list(data.keys())
print(f"总航班数（传感器数据）: {len(flight_ids)}")
print(f"总航班数（标签数据）: {len(header_df)}")

# 3. 检查哪些航班ID在两边都存在
common_ids = set(flight_ids) & set(header_df['Master Index'].values)
print(f"共同存在的航班数: {len(common_ids)}")

# 4. 提取第一个航班的数据作为示例
sample_id = list(common_ids)[0]
sample_data = data[sample_id]
sample_label = header_df[header_df['Master Index'] == sample_id]['before_after'].values[0]
print(f"\n示例航班 ID: {sample_id}")
print(f"传感器数据形状: {sample_data.shape}")  # (时间步长, 传感器数量)
print(f"标签 (before_after): {sample_label}")  # 0 或 1
```

```python
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from sktime.transformations.panel.rocket import MiniRocketMultivariate
from sklearn.linear_model import RidgeClassifierCV
from sklearn.pipeline import Pipeline
import time

# 1. 准备数据：将字典转换为 X (特征列表) 和 y (标签)
X = []  # 每个元素是一个 (时间步长, 传感器数) 的 numpy 数组
y = []  # 对应的标签

print("正在准备数据...")
for flight_id in common_ids:
    X.append(data[flight_id])  # 原始传感器数据
    label = header_df[header_df['Master Index'] == flight_id]['before_after'].values[0]
    y.append(label)
X = np.array(X, dtype=object)  # 注意：dtype=object 是因为每个数组形状不同
y = np.array(y)
print(f"X 长度: {len(X)}, y 长度: {len(y)}")
print(f"标签分布: 0={sum(y==0)}, 1={sum(y==1)}")

# 2. 定义 MiniRocket 管道
pipeline = Pipeline([
    ("minirocket", MiniRocketMultivariate(random_state=42)),
    ("classifier", RidgeClassifierCV(alphas=np.logspace(-3, 3, 10)))
])

# 3. 五折交叉验证
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
accuracies = []
f1_scores = []
auc_scores = []

print("\n开始五折交叉验证...")
fold = 1
for train_idx, val_idx in cv.split(X, y):
    print(f"\n--- Fold {fold} ---")
    X_train, X_val = [X[i] for i in train_idx], [X[i] for i in val_idx]
    y_train, y_val = y[train_idx], y[val_idx]
    
    # 训练
    start = time.time()
    pipeline.fit(X_train, y_train)
    train_time = time.time() - start
    
    # 预测
    y_pred = pipeline.predict(X_val)
    y_prob = pipeline.predict_proba(X_val)[:, 1]
    
    # 评估
    acc = accuracy_score(y_val, y_pred)
    f1 = f1_score(y_val, y_pred)
    auc = roc_auc_score(y_val, y_prob)
    
    accuracies.append(acc)
    f1_scores.append(f1)
    auc_scores.append(auc)
    
    print(f"准确率: {acc:.4f}, F1: {f1:.4f}, AUC: {auc:.4f}, 训练时间: {train_time:.2f}s")
    fold += 1

# 4. 报告最终结果
print("\n" + "="*50)
print("五折交叉验证最终结果:")
print(f"准确率: {np.mean(accuracies):.4f} ± {np.std(accuracies):.4f}")
print(f"F1分数: {np.mean(f1_scores):.4f} ± {np.std(f1_scores):.4f}")
print(f"AUC: {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("="*50)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from sktime.transformations.panel.rocket import MiniRocketMultivariate
from sklearn.linear_model import LogisticRegressionCV
import time

# 1. 准备数据
print("正在准备数据...")
max_length = 0
for flight_id in common_ids:
    sensor_data = data[flight_id]
    if sensor_data.shape[0] > max_length:
        max_length = sensor_data.shape[0]
print(f"最长序列长度: {max_length}")

X_list = []
y = []
for flight_id in common_ids:
    sensor_data = data[flight_id]
    sensor_data = np.nan_to_num(sensor_data, nan=0.0)
    if sensor_data.shape[0] < max_length:
        pad_width = ((0, max_length - sensor_data.shape[0]), (0, 0))
        sensor_data = np.pad(sensor_data, pad_width, mode='constant', constant_values=0)
    X_list.append(sensor_data)
    label = header_df[header_df['Master Index'] == flight_id]['before_after'].values[0]
    y.append(label)

X = np.array(X_list)
y = np.array(y)
print(f"X 形状: {X.shape}")
print(f"y 长度: {len(y)}")
print(f"标签分布: 0={sum(y==0)}, 1={sum(y==1)}")

# 2. 初始化
minirocket = MiniRocketMultivariate(random_state=42)
# 改用 LogisticRegressionCV，能输出概率
classifier = LogisticRegressionCV(
    Cs=10, 
    cv=5, 
    random_state=42,
    max_iter=1000
)

# 3. 五折交叉验证
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
accuracies = []
f1_scores = []
auc_scores = []

print("\n开始五折交叉验证...")
fold = 1
for train_idx, val_idx in cv.split(X, y):
    print(f"\n--- Fold {fold} ---")
    X_train, X_val = X[train_idx], X[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]
    
    start = time.time()
    X_train_transform = minirocket.fit_transform(X_train, y_train)
    classifier.fit(X_train_transform, y_train)
    X_val_transform = minirocket.transform(X_val)
    y_pred = classifier.predict(X_val_transform)
    y_prob = classifier.predict_proba(X_val_transform)[:, 1]
    train_time = time.time() - start
    
    acc = accuracy_score(y_val, y_pred)
    f1 = f1_score(y_val, y_pred)
    auc = roc_auc_score(y_val, y_prob)
    
    accuracies.append(acc)
    f1_scores.append(f1)
    auc_scores.append(auc)
    
    print(f"准确率: {acc:.4f}, F1: {f1:.4f}, AUC: {auc:.4f}, 训练时间: {train_time:.2f}s")
    fold += 1

print("\n" + "="*50)
print("五折交叉验证最终结果:")
print(f"准确率: {np.mean(accuracies):.4f} ± {np.std(accuracies):.4f}")
print(f"F1分数: {np.mean(f1_scores):.4f} ± {np.std(f1_scores):.4f}")
print(f"AUC: {np.mean(auc_scores):.4f} ± {np.std(auc_scores):.4f}")
print("="*50)
```
